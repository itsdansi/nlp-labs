# Exercise 11: Predict the Next Word Using Conditional Probability

Use bigram counts to compute P(w2 | w1) and predict the most likely next word.

## Background

The conditional probability of a word w2 given a previous word w1 is:

P(w2 | w1) = Count(w1, w2) / Count(w1)

where Count(w1, w2) is the number of times the bigram appears, and Count(w1) is how many times w1 appears as the first word of a bigram.

## Step 1: Read the training corpus

```python
corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP",
    "You like coffee"
]

print("Training corpus:")
for sentence in corpus:
    print(sentence)
```

## Step 2: Build bigram counts

```python
bigram_counts = {}
unigram_as_first = {}

for sentence in corpus:
    words = sentence.split()
    for i in range(len(words) - 1):
        w1 = words[i]
        w2 = words[i + 1]

        if w1 not in bigram_counts:
            bigram_counts[w1] = {}
        if w2 not in bigram_counts[w1]:
            bigram_counts[w1][w2] = 0
        bigram_counts[w1][w2] += 1

        if w1 not in unigram_as_first:
            unigram_as_first[w1] = 0
        unigram_as_first[w1] += 1

print("Bigram Count")
print(f"{'Bigram':<18} Count")
for w1, next_words in bigram_counts.items():
    for w2, count in next_words.items():
        print(f"{w1} -> {w2:<10} {count}")
```

## Step 3: Calculate conditional probabilities for every bigram

```python
conditional_probabilities = {}

print("Conditional probabilities P(w2 | w1):\n")
for w1, next_words in bigram_counts.items():
    conditional_probabilities[w1] = {}
    total = unigram_as_first[w1]
    print(f"After '{w1}' (total = {total}):")
    for w2, count in next_words.items():
        probability = count / total
        conditional_probabilities[w1][w2] = probability
        print(f"  P({w2} | {w1}) = {count}/{total} = {probability:.2f}")
    print()
```

## Worked examples from the sheet

like is followed by NLP (1), Python (1), and coffee (2), so the total is 4.

- P(coffee | like) = 2/4 = 0.50
- P(Python | like) = 1/4 = 0.25
- P(NLP | like) = 1/4 = 0.25

## Step 4–6: Ask for a word, show next-word probabilities, and predict

```python
# Sample input from the assignment. Change this line to input("Enter a word: ") to type a word.
user_word = "like"
print(f"Enter a word: {user_word}")

if user_word not in conditional_probabilities:
    print("This word never appears as the first word of a bigram.")
else:
    next_words = conditional_probabilities[user_word]
    ranked = sorted(next_words.items(), key=lambda item: item[1], reverse=True)

    print("Possible next words:")
    for word, probability in ranked:
        print(f"{word} : {probability:.2f}")

    best_word, best_probability = ranked[0]
    print(f"\nPredicted next word: {best_word}")
```

## Bonus: Top three predictions

```python
user_word = "like"

if user_word in conditional_probabilities:
    ranked = sorted(
        conditional_probabilities[user_word].items(),
        key=lambda item: item[1],
        reverse=True
    )
    top_three = ranked[:3]

    print("Top Predictions")
    for index, (word, probability) in enumerate(top_three, start=1):
        print(f"{index}. {word} ({probability:.2f})")
```
