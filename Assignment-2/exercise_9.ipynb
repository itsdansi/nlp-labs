# Exercise 9: Find Similar Words

Find the word most similar to a given word using cosine similarity on embeddings.

##  Store the embeddings

Use the given word embeddings as dense vectors.

```python
embeddings = {
    "cat": [0.4, 0.2, 0.8],
    "dog": [0.5, 0.1, 0.7],
    "car": [-0.7, 0.8, -0.2],
    "lion": [0.45, 0.25, 0.82]
}

print("Embeddings loaded for:")
for word in embeddings:
    print(word)
```

## Define cosine similarity

Create a helper function to compare two vectors.

```python
import math

def cosine_similarity(vector_a, vector_b):
    dot_product = sum(a * b for a, b in zip(vector_a, vector_b))
    magnitude_a = math.sqrt(sum(a * a for a in vector_a))
    magnitude_b = math.sqrt(sum(b * b for b in vector_b))
    return dot_product / (magnitude_a * magnitude_b)
```

## Compare the target word with all others

Find the word most similar to `cat`.

```python
target_word = "cat"
target_vector = embeddings[target_word]

similarities = {}


def find_most_similar_word(target_word):
    target_vector = embeddings[target_word]
    similarities = {}

    for word, vector in embeddings.items():
        if word != target_word:
            similarities[word] = cosine_similarity(target_vector, vector)

    most_similar_word = max(similarities, key=similarities.get)
    most_similar_score = similarities[most_similar_word]

    print(f"Most similar to {target_word}")
    print(most_similar_word)
    print(f"Similarity = {most_similar_score:.2f}")
    print()

for target_word in ["cat", "dog", "car"]:
    find_most_similar_word(target_word)
```
