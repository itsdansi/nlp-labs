# Q4. Detect data sparsity

Find word pairs that never occur together using the co-occurrence matrix from Exercise 3.

## Co-occurrence matrix (from Exercise 3)

```python
sentence = "the cat sat on the mat"
words = sentence.split()

vocab = list(dict.fromkeys(words))
word_to_index = {word: i for i, word in enumerate(vocab)}

size = len(vocab)
matrix = [[0] * size for _ in range(size)]

for i, word in enumerate(words):
    current = word_to_index[word]

    if i > 0:
        left = word_to_index[words[i - 1]]
        matrix[current][left] += 1

    if i < len(words) - 1:
        right = word_to_index[words[i + 1]]
        matrix[current][right] += 1

print("    ", *vocab)
for word, row in zip(vocab, matrix):
    print(f"{word:4}", *row)
```

## Word pairs that never occurred

Traverse the matrix and print every off-diagonal cell whose value is `0`.

```python
for i, row_word in enumerate(vocab):
    for j, col_word in enumerate(vocab):
        if i != j and matrix[i][j] == 0:
            print(f"({row_word}, {col_word}) -> Never occurred")
```
