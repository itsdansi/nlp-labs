# Exercise 8: Dense Vector Similarity

Compute cosine similarity between two dense vectors.

## Define the vectors

Store the two dense vectors that we will compare.

```python
cat = [0.4, 0.2, 0.8]
dog = [0.5, 0.1, 0.7]

print("cat vector:", cat)
print("dog vector:", dog)
```

##  Compute the dot product

Multiply matching positions and add the results.

```python
dot_product = sum(a * b for a, b in zip(cat, dog))
print(f"Dot Product = {dot_product:.4f}")
```

##  Compute vector magnitudes

Find the length of each vector using the square root of the sum of squares.

```python
import math

cat_magnitude = math.sqrt(sum(value * value for value in cat))
dog_magnitude = math.sqrt(sum(value * value for value in dog))

print(f"cat magnitude = {cat_magnitude:.4f}")
print(f"dog magnitude = {dog_magnitude:.4f}")
```

## Calculate cosine similarity

Divide the dot product by the product of the magnitudes.

```python
cosine_similarity = dot_product / (cat_magnitude * dog_magnitude)
print(f"Cosine Similarity = {cosine_similarity:.2f}")
```
