# Q. Count how many times each word appears in a sentence.

```python
text = "I like NLP I like Python NLP is fun"
words = text.split()
```

## Manual counting

```python
word_count = {}

for w in words:
    if w in word_count:
        word_count[w] += 1
    else:
        word_count[w] = 1

print(word_count)
```

## Using `Counter`

```python
from collections import Counter

word_count = Counter(words)
print(word_count)
```
