# Exercise 7: Calculate TF-IDF

Compute TF, IDF, and TF-IDF scores for a small corpus.

## Prepare the corpus

Store the documents in a list for processing.

```python
documents = [
    "I like NLP",
    "I like AI",
    "AI is amazing"
]

print("Corpus:")
for index, document in enumerate(documents, start=1):
    print(f"Document {index}: {document}")
```

## Calculate term frequency

Count how often each word appears in each document and divide by the total number of words in that document.

```python
from collections import Counter

tokenized_documents = [document.split() for document in documents]
term_frequencies = []

for tokens in tokenized_documents:
    counts = Counter(tokens)
    total_terms = len(tokens)
    tf_scores = {term: count / total_terms for term, count in counts.items()}
    term_frequencies.append(tf_scores)

for index, tf_scores in enumerate(term_frequencies, start=1):
    print(f"TF for Document {index}: {tf_scores}")
```

## Calculate document frequency

Count in how many documents each term appears.

```python
import math

document_frequency = Counter()
all_terms = set()

for tokens in tokenized_documents:
    unique_terms = set(tokens)
    all_terms.update(unique_terms)
    for term in unique_terms:
        document_frequency[term] += 1

print("DF values:")
for term in sorted(all_terms):
    print(f"{term} : {document_frequency[term]}")
```

## Calculate IDF

Use the natural logarithm formula: `idf = ln(total_documents / df)`.

```python
total_documents = len(documents)
idf_scores = {term: math.log(total_documents / document_frequency[term]) for term in all_terms}

print("IDF values:")
for term in sorted(idf_scores):
    print(f"{term} : {idf_scores[term]:.4f}")
```

## Compute TF-IDF

Multiply TF by IDF for every term in each document.

```python
tf_idf_scores = []

for tf_scores in term_frequencies:
    document_scores = {term: tf_scores.get(term, 0) * idf_scores[term] for term in sorted(all_terms)}
    tf_idf_scores.append(document_scores)

for index, scores in enumerate(tf_idf_scores, start=1):
    print(f"TF-IDF for Document {index}:")
    for term, score in scores.items():
        print(f"  {term} : {score:.4f}")
```
