# Exercise 5: Implement Laplace Smoothing

Calculate word probabilities using Laplace smoothing.

## Store the word counts

Start with the given word counts in a dictionary.

```python
word_counts = {
    "cat": 4,
    "dog": 3,
    "bird": 1
}
```

## Task 1: Compute the total number of words

Add up all the counts to get the total number of words.

```python
total_words = sum(word_counts.values())
print(f"Total words: {total_words}")
```

## Task 2: Compute the vocabulary size

Count how many unique words are in the dictionary.

```python
vocabulary_size = len(word_counts)
print(f"Vocabulary size: {vocabulary_size}")
```

## Task 3: Apply Laplace smoothing

Use the formula `(count + 1) / (total_words + vocabulary_size)` for each word.

```python
laplace_probabilities = {}
for word, count in word_counts.items():
    probability = (count + 1) / (total_words + vocabulary_size)
    laplace_probabilities[word] = probability
    print(f"{word} : {probability:.2f}")
```
