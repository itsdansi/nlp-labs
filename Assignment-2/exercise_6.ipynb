# Exercise 6: Simple Keyboard Prediction

Predict the most common next word using bigram counts.

## Prepare the training corpus

Use the given sentences to build a simple bigram language model.

```python
training_corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP"
]

print("Training corpus:")
for sentence in training_corpus:
    print(sentence)
```

## Build the bigram counts

Count how often each pair of words appears in the corpus.

```python
from collections import defaultdict

bigram_counts = defaultdict(lambda: defaultdict(int))

for sentence in training_corpus:
    words = sentence.split()
    for current_word, next_word in zip(words, words[1:]):
        bigram_counts[current_word][next_word] += 1

for current_word, next_words in bigram_counts.items():
    print(f"{current_word} -> {dict(next_words)}")
```

## Predict the next word

Use the input phrase `I like` and show the three most likely next words.

```python
input_phrase = "I like"
last_word = input_phrase.split()[-1]

next_word_frequencies = bigram_counts[last_word]
top_predictions = sorted(
    next_word_frequencies.items(),
    key=lambda item: (-item[1], item[0])
)[:3]

print("Suggestions")
for word, _ in top_predictions:
    print(word)
```
