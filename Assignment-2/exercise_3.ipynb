# Q3. Build a word co-occurrence matrix (window size = 1).

```python
sentence = "the cat sat on the mat"

words = sentence.split()
```

## Build vocabulary (preserve order)

```python
vocab = list(dict.fromkeys(words))
word_to_index = {word: i for i, word in enumerate(vocab)}
print(vocab)
print(word_to_index)
```

## Co-occurrence matrix

```python
size = len(vocab)
matrix = [[0] * size for _ in range(size)]

for i, word in enumerate(words):
    current = word_to_index[word]

    # Left neighbor
    if i > 0:
        left = word_to_index[words[i - 1]]
        matrix[current][left] += 1

    # Right neighbor
    if i < len(words) - 1:
        right = word_to_index[words[i + 1]]
        matrix[current][right] += 1

print("    ", *vocab)

for word, row in zip(vocab, matrix):
    print(f"{word:4}", *row)
```
