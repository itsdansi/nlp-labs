```python
from pathlib import Path
import csv
import nltk
from nltk import pos_tag, word_tokenize

for resource in ("punkt_tab", "averaged_perceptron_tagger_eng"):
    nltk.download(resource, quiet=True)

raw_text = Path("News_Article.txt").read_text(encoding="utf-8")
text = raw_text.split("\n\n", 1)[-1].rsplit("\n\nhttp", 1)[0].strip()
tagged_words = pos_tag(word_tokenize(text))
print(f"Tagged {len(tagged_words)} tokens.")
```

```python
rows = [
    (word, tag)
    for word, tag in tagged_words
    if word.isalpha() and tag.startswith(("NN", "VB"))
]

with open("DansiAcharya_POS_01.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Word", "POS_Tag"])
    writer.writerows(rows)

print(f"Saved {len(rows)} nouns and verbs to DansiAcharya_POS_01.csv")
rows[:15]
```
