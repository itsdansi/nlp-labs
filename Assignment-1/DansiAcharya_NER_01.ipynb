# Assignment 02: Named Entity Recognition

Library: **NLTK**  
Source: [The Kathmandu Post](https://kathmandupost.com/national/2026/07/08/government-appoints-govinda-bahadur-karki-as-chief-secretary) (July 8, 2026).

```python
from collections import Counter
from pathlib import Path
import csv
import nltk
from nltk import ne_chunk, pos_tag, sent_tokenize, word_tokenize
from nltk.tree import Tree

for resource in (
    "punkt_tab",
    "averaged_perceptron_tagger_eng",
    "maxent_ne_chunker_tab",
    "words",
):
    nltk.download(resource, quiet=True)

raw_text = Path("News_Article.txt").read_text(encoding="utf-8")
text = raw_text.split("\n\n", 1)[-1].rsplit("\n\nhttp", 1)[0].strip()
print(f"Loaded {len(word_tokenize(text))} tokens.")
```

```python
entities = []
for sentence in sent_tokenize(text):
    tree = ne_chunk(pos_tag(word_tokenize(sentence)), binary=False)
    for node in tree:
        if isinstance(node, Tree):
            entity = " ".join(word for word, _ in node.leaves())
            entities.append((entity, node.label()))

with open("DansiAcharya_NER_01.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Entity", "Entity_Type"])
    writer.writerows(entities)

print(f"Saved {len(entities)} entities to DansiAcharya_NER_01.csv")
entities
```

```python
summary = Counter(label for _, label in entities)
print("Entity type summary")
for label, count in sorted(summary.items()):
    print(f"{label}: {count}")
```
